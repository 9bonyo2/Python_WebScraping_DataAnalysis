##### 질문2 : 여러개의 section 중 하나를 선택해서 url에서 뉴스기사의 링크와 제목을 출력하는 코드를 함수로 작성하기

```python
# reqeusts, bs4 import
import requests
import bs4
# BeautifulSoup 클래스 import
from bs4 import BeautifulSoup

#각 section별 sid
section_dict={'기후/환경':'climate','사회':'society','경제':'economy','정치':'politics','국제':'world','문화':'culture', '생활':'life', 'IT/과학':'tech','인물':'people','지식/칼럼':'understanding','연재':'series'}


def print_news(section_name):  #print_new('기후/환경') 
    # 검색시 키워드 일부만 검색해도 검색이 가능하도록 구현 (ex. 기후로 검색 -> 기후/환경 뉴스로 표시.)
    # 입력한 단어가 section 이름에 포함되어 있는지 확인
    for name in section_dict:
        if section_name in name:
            section_name = name
            break

    sid = section_dict.get(section_name,'climate')

    # 존재하지 않는 section이면 종료
    if sid is None:
        print('존재하지 않는 뉴스 섹션입니다.')
        return

    url = f'https://news.daum.net/{sid}'
    print(f'{section_name} 뉴스 {url}')


    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/134.0.0.0 Safari/537.36'
    }

    res=requests.get(url, headers=req_header)
    # print(res.encoding)
    # print(res.apparent_encoding)
    # print(res.headers.get('content-type'))

    if res.ok:    
        # 응답(response)에서 text 추출
        res.encoding = 'utf-8'    
        html=res.text
        soup=BeautifulSoup(html, 'html.parser')
        #print(html)

        #기사 제목과 url 추출
        # print(len(soup.select("ul.list_newsheadline2 li")))
        li_tags=soup.select("ul.list_newsheadline2 li")
        # print(li_tags)
        # print(type(li_tags))

    # <li> 태그 리스트 순회하기    
        for li_tag in li_tags:
            #print(li_tags)
            
            #뉴스 기사 제목
            strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
            title = strong_tag.text
            # print(title)

            #뉴스 기사 url
            a_tag = li_tag.find('a')
            link = a_tag['href']
            # print(link)

            print(title, link)
            
            
    else:
        print(f'Error Code = {res.status_code}')



    
```

### 2-1. Nate 뉴스기사 제목 스크래핑하기
#### 출력 내용 : 최신뉴스, 정치 , 경제, 사회, 세계, IT/과학 6개의 섹션의 뉴스를 출력하는 함수를 생성하여 스크래핑 하기
##### Image, 기사제목, 기사링크
##### 1) Image의 도메인이름이 포함된 url과 src 속성의 img 경로를 합치려면 urljoin 함수를 사용하세요.  
###### from urllib.parse import urljoin    
###### url = ‘https://news.nate.com/recent?mid=n0100’    
###### src=’ //thumbnews.nateimg.co.kr/news90///news.nateimg.co.kr/orgImg/na/2025/07/23/7408335_high.jpg’ 
##### 2) Image 출력은 Image 클래스와 display 함수를 사용하세요.
###### from IPython.display import Image, display 
##### 3) img 엘리먼트의 존재 여부를 체크하신 후에 src 속성의 이미지를 경로를 추출하기   
###### => Image 가 없는 뉴스도 있기 때문에



```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin #URL 합치기 위한 라이브러리
from IPython.display import Image, display #쥬피터 노트북 화면에 이미지를 보여주기 위한 라이브러리


# 각 뉴스 section별 mid
section_dict = { '최신뉴스': 'n0100', '정치': 'n0101', '경제': 'n0102', '사회': 'n0103', '세계': 'n0104', 'IT/과학': 'n0105'}

def print_nate_news(section_name):

    # section_name에 해당하는 mid 가져오기
    mid = section_dict.get(section_name)

    # 요청 URL
    url = f'https://news.nate.com/recent?mid={mid}'

    print(f'{section_name} 뉴스')
    print(url)


    # 요청 헤더 설정
    req_header ={ 
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
    }


    # 뉴스 페이지 요청
    res = requests.get(url, headers=req_header)

    if res.ok:
        # BeautifulSoup 객체 생성
        soup = BeautifulSoup(res.text, 'html.parser')

        # 뉴스 기사 목록
        news_list = soup.select('div.mduSubjectList')

        print(f'기사 개수 : {len(news_list)}')

        # 뉴스 기사 순회
        for news in news_list:

            #print(news_list)

            # 뉴스 기사 제목
            h2_tag = news.find('h2')
            #print(h2_tag)
            title=h2_tag.text.strip()

            # 뉴스 기사 링크
            a_tag = news.find('a')
            link = urljoin(url, a_tag['href'])


            # 뉴스 이미지
            img_tag = news.find('img')

            # img 태그가 존재하는지 확인
            if img_tag is not None and img_tag.get('src'):

                # 이미지 src 가져오기
                img_src = img_tag['src']

                # 이미지 전체 URL 만들기
                img_url = urljoin(url, img_src)

                # 이미지 출력
                display(Image(url=img_url, width=200))

                print('Image :', img_url)

            else:
                print('Image : 없음')


            print('기사제목 :', title)
            print('기사링크 :', link)
            

    else:
        print(f'Error Code = {res.status_code}')

print_nate_news('경제')
    
```

#### 2-2. 하나의 네이버 웹툰과 1개의 회차에 대한 Image 다운로드 하기 (필수)

```python
import requests
from bs4 import BeautifulSoup
import re #정규표현식 사용


def download_image(title, no, url):
    import requests
    from bs4 import BeautifulSoup
    import os

    # 이미지 요청 시 사용할 referer
    req_header = {
        'referer': url
    }

    # 웹툰 회차 페이지 요청
    res = requests.get(url)
    print(res.ok)

    if res.ok:
        soup = BeautifulSoup(res.text, 'html.parser')
        
        imgurl_list = []
        img_tags = soup.select("img[src*='IMAG01']")
        #print(len(soup.select("img[src*='IMAG01']")))
        print(f'이미지 개수 : {len(img_tags)}')

        #이미지 URL 추출
        for img_tag in img_tags:
            imgurl_list.append(img_tag['src'])
            #print(img_tag)
                
        #image를 저장할 디렉토리 생성
        #img/일렉시드/341
        #dir_name = 'img/' + title + no
        dir_name = os.path.join('img', title, str(no))
        #print(dir_name)
        print(f'저장경로 : {dir_name}')

        # 디렉토리가 없으면 생성
        if not os.path.isdir(dir_name):
            os.makedirs(dir_name)

        #이미지 다운로드
        for idx,img_url in enumerate(imgurl_list,1):
            #print(idx, img_url)
            print(f'---> 다운로드 번호 {idx}')
            res = requests.get(img_url,headers=req_header)
            #print(res.status_code)

            if res.ok:
                # 이미지 binary 데이터
                img_data = res.content
                #print(len(img_data))

                # URL에서 파일명 가져오기
                file_name = os.path.basename(img_url)
                #print(file_name)

                # 저장할 전체 경로
                file_name = dir_name + '/' + file_name

                # 이미지 파일 저장
                with open(file_name,'wb') as file:
                    print(f'파일명 : {file_name} (파일크기 : {len(img_data)}))')
                    file.write(img_data)
                
                
download_image('일렉시드',341,'https://comic.naver.com/webtoon/detail?titleId=717481&no=341&week=wed')
```
